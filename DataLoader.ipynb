```python
import yfinance as yf
from langchain_openai import ChatOpenAI
import os
api_key = os.getenv("OPENAI_API_KEY")
```

```python
class DataLoader:
    def __init__(self,ticker,from_date,to_date):
        self.ticker = ticker
        self.from_date = from_date
        self.to_date = to_date
        print("Variables initialised")
    def fetch_data(self):
        try:
            data = yf.download(self.ticker,start= self.from_date,end=self.to_date)
            if data.empty:
                raise ValueError("No Data Fetched")
            return data
        except Exception as e:
            raise e
    def clean_data(self,data):
        #remove any null or NaN values
        data = data.dropna()
        #Making sure the data is ordered by date
        data = data.sort_index(ascending = True)
    def get_data(self):
        data = self.fetch_data()
        cleaned_data = self.clean_data(data)
        return cleaned_data
```

```python
loader = DataLoader("AAPL","2000-04-01","2001-04-01")
```

```python
app = loader.get_data()
```

```python
loader.clean_data(app)
```
